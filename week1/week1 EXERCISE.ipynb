```python
# imports
from openai import OpenAI
import os
from dotenv import load_dotenv
from IPython.display import display, Markdown, update_display
```

```python
# constants

MODEL_GEMINI = 'gemini-3.5-flash-lite'
MODEL_LLAMA = 'llama3.2:1b'
```

```python
# set up environment

# Gemini enviroment
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

load_dotenv(override=True)

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("No API key was found")
elif not google_api_key.startswith(("AIz", "AQ.")):
    print("An API key was found, but it doesn't start with AIz or AQ.")
else:
    print("API key found and looks good so far.")

openai = OpenAI(
    base_url=GEMINI_BASE_URL,
    api_key= google_api_key
)
```

```python
#Ollama enviroment

OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
```

```python
system_prompt = """
You are a ex football player and an expert in International Football, who answer with homourous answers. 
"""
```

```python
# here is the question; type over this to ask something new

question = """
Please, answer the following question: 
Which football player is the best in the world? 
"""
```

```python
# Get gemini-3.5-flash-lite to answer, with streaming
def stream_answer(prompt, user_promt):
    stream = openai.chat.completions.create(
        model="gemini-3.5-flash",
        messages=[
            {"role": "system", "content": prompt},
            {"role": "user", "content":user_promt}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream: 
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
    

stream_answer(system_prompt, question)
```

```python
# Get Llama 3.2 to answer

def stream_answer_ollama(prompt, user_prompt):
    stream = ollama.chat.completions.create(
        model="llama3.2:1b",
        messages=[
            {"role": "system", "content":prompt},
            {"role": "user", "content": user_prompt}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream: 
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

stream_answer(system_prompt, question)
```
